#### 6.Sum of weights

```python
x = df[df['origin'] == "Asia"][['vehicle_weight','model_year']].head(7)
XTX = x.T.dot(x)
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
XTXiXT = XTX_inv.dot(x.T)
w = XTXiXT.dot(y)
w.sum()
```
